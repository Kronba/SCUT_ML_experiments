```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import auc, precision_recall_curve, roc_curve

plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.facecolor'] = 'white'
plt.rcParams['axes.facecolor'] = 'white'
plt.rcParams['savefig.facecolor'] = 'white'
plt.rcParams['text.color'] = 'black'
plt.rcParams['axes.labelcolor'] = 'black'
plt.rcParams['xtick.color'] = 'black'
plt.rcParams['ytick.color'] = 'black'
plt.rcParams['grid.color'] = '#e0e0e0'
```

```python
labels = np.array([1, 1, 0, 1, 0, 1, 0, 0])
scores = np.array([0.81, 0.74, 0.62, 0.53, 0.44, 0.35, 0.25, 0.21])
```

```python
precision, recall, thresholds = precision_recall_curve(labels, scores)
```

```python
print('--- P-R Curve Endpoints (Recall, Precision) ---')
for r, p in zip(recall, precision):
  print(f'({r:.2f}, {p:.2f})')

plt.figure(figsize=(6, 5))
plt.plot(
    recall,
    precision,
    marker='o',
    drawstyle='steps-post',
    color='blue',
    label='P-R Curve',
    linewidth=2,
    markersize=6
)
plt.xlabel('Recall', fontsize=12, fontweight='bold')
plt.ylabel('Precision', fontsize=12, fontweight='bold')
plt.title('Precision-Recall Curve', fontsize=14, fontweight='bold')
plt.xlim([-0.05, 1.05])
plt.ylim([-0.05, 1.05])
plt.grid(True, linestyle='--', alpha=0.5)
# Ensure legend background is white
legend = plt.legend(loc='lower left', frameon=True)
frame = legend.get_frame()
frame.set_facecolor('white')
frame.set_edgecolor('black')
for text in legend.get_texts():
    text.set_color('black')
plt.tight_layout()
plt.show()
```

```python
fpr, tpr, roc_thresholds = roc_curve(labels, scores)
roc_auc = auc(fpr, tpr)
```

```python
print('--- ROC Curve Endpoints (FPR, TPR) ---')
for f, t in zip(fpr, tpr):
  print(f'({f:.2f}, {t:.2f})')
print(f'\nAUC Value = {roc_auc:.4f}')

plt.figure(figsize=(6, 5))
plt.plot(
    fpr,
    tpr,
    marker='o',
    color='red',
    label=f'ROC Curve (AUC = {roc_auc:.4f})',
    linewidth=2,
    markersize=6
)
plt.plot([0, 1], [0, 1], color='gray', linestyle='--', label='Random Guess', linewidth=1.5)
plt.xlabel('False Positive Rate (FPR)', fontsize=12, fontweight='bold')
plt.ylabel('True Positive Rate (TPR)', fontsize=12, fontweight='bold')
plt.title('Receiver Operating Characteristic (ROC) Curve', fontsize=14, fontweight='bold')
plt.xlim([-0.05, 1.05])
plt.ylim([-0.05, 1.05])
plt.grid(True, linestyle='--', alpha=0.5)
# Ensure legend background is white
legend = plt.legend(loc='lower right', frameon=True)
frame = legend.get_frame()
frame.set_facecolor('white')
frame.set_edgecolor('black')
for text in legend.get_texts():
    text.set_color('black')
plt.tight_layout()
plt.show()
```
